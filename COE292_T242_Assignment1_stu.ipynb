# 🚀 Task 2: Implement K-Means Clustering from Scratch 🚀

```python
import numpy as np

# ==========================================
# 🔹 Instructions:
# - Complete the missing code in each function.
# - Do NOT modify the provided function structure, input parameters, or return statements.
# ==========================================

def mykmeans(df, n_clusters=4, max_iter=1000): # (❌ DO NOT MODIFY the function header, you may try different values of max_iter for convergence)
    """
    K-Means clustering

    Parameters:
    - df: Dataset (numpy array) (Already randomized)
    - n_clusters: Number of clusters to make (Your solution should work for any provided reasonable value of n_clusters and not only for a specific value of n_clusters)
    - max_iter: Maximum number of iterations (default=1000)

    Returns:
    - centroids (numpy array): Final cluster centroids after the KMeans clustering
    """

    centroids =  get_initial_centroids(df, n_clusters) # (❌ DO NOT MODIFY) (This is to get randomly selected initial centroids)

    for _ in range(max_iter):

        # ------------------ Your Code Here ------------------
        #we first have to find the distances of every point to the centroid
        #put them in a list so we can come back to them
        classifications = [[] for _ in range(n_clusters)]
        distances= []
        for i in range(len(df)):
              distances = []
              for c in centroids:
                  distance = get_euclidean(df[i], c) # The distance to each centroid
                  distances.append(distance)
              classification = np.argmin(distances)
              classifications[classification].append(df[i]) #Minimum distance


        updated = np.zeros((n_clusters, df.shape[1])) #

        for i in range(n_clusters):
            if classifications[i]:
                updated[i] = np.mean(classifications[i], axis=0)

        if np.all(centroids == updated):
            break
        centroids = updated

        # ---------------------------------------------------

    centroids = centroids[np.argsort(centroids[:, 0])]       # ❌ DO NOT MODIFY
    return centroids  # ❌ DO NOT MODIFY this return statement

```

## 📌 How to find WCSS?

Within-Cluster-Sum of Squared Errors can be computed using the following steps:

 - The Squared Error for each point is the square of the distance of the point from its closest cluster centroid.
 - The WCSS score is the sum of these Squared Errors for all the points
 - The distance metric could be any distance function such as Euclidean Distance and/or the Manhattan Distance
 - A generic formula for WCSS = ∑(d(point, centroid))²  , where `d(.,.)` will be replaced by the euclidean or manhattan distance function specified below:
    

  Euclidean Distance:

   - $ \text{d(A,B)} =  \sqrt{ \sum_{i} (A_i - B_i)^2  }  $,  (where A_i is the i-th element (or item) in vector A)

  Manhattan Distance:

   - $ \text{d(A,B)} =   \sum_{i} |A_i - B_i|  $,    (where A_i is the i-th element (or item) in vector A)




```python
import numpy as np


def get_wcss(df, centroids): # ❌ DO NOT MODIFY the function arguments
    """
    Compute the Within-Cluster Sum of Squares (WCSS) using both Euclidean and Manhattan distances.

    :param df: Dataset (numpy array), where each row represents a data point.
    :param centroids: Cluster centroids (numpy array).
    :return: A tuple (WCSS_Euclidean, WCSS_Manhattan).
    """

    # ----------------------------- Your Code Here ------------------------------- #

    wcss_euclidean = 0
    wcss_manhattan = 0
    for i in range(len(df)):# Iterating in each point
        euMin, manMin = float('inf'), float('inf') #Initilizing

        for c in centroids:# The distances for each centroid
            euMetric = get_euclidean(df[i], c)
            manMetric = get_manhattan(df[i], c)

            if euMetric < euMin:
                euMin = euMetric
            if manMetric < manMin:
                manMin = manMetric

        wcss_euclidean += euMin ** 2 #Applying formula
        wcss_manhattan += manMin ** 2

    return wcss_euclidean, wcss_manhattan  # ❌ DO NOT MODIFY this return statement

    # ---------------------------------------------------------------------------- #


    return wcss_euclidean, wcss_manhattan  # ❌ DO NOT MODIFY this return statement
```
